# Tianchi News Recommendation - Solution Comparison

This notebook runs all 4 solution pipelines and compares their performance.

| Solution | Recall | Ranking |
|----------|--------|--------|
| S1: news_recommender | ItemCF + YouTubeDNN | None (recall-only) |
| S2: top2 | ItemCF + Bi-network + W2V | LightGBM Classifier |
| S3: top3 | Hot + 30s-i2i | LightGBM Ranker |
| S5: top5 | ItemCF + Swing + Item2Vec | LightGBM Ranker |

```python
import sys
sys.path.insert(0, '..')

import warnings
warnings.filterwarnings('ignore')
```

```python
from solutions.solution1_news_recommender import Solution1NewsRecommender
from solutions.solution2_top2 import Solution2Top2
from solutions.solution3_top3 import Solution3Top3
from solutions.solution5_top5 import Solution5Top5
```

```python
# Run Solution 1: news_recommender
s1 = Solution1NewsRecommender()
metrics_s1 = s1.run()
```

```python
# Run Solution 2: top2
s2 = Solution2Top2()
metrics_s2 = s2.run()
```

```python
# Run Solution 3: top3
s3 = Solution3Top3()
metrics_s3 = s3.run()
```

```python
# Run Solution 5: top5
s5 = Solution5Top5()
metrics_s5 = s5.run()
```

```python
# Comparison table
import pandas as pd

all_metrics = {
    'S1: news_recommender': metrics_s1,
    'S2: top2': metrics_s2,
    'S3: top3': metrics_s3,
    'S5: top5': metrics_s5,
}

df = pd.DataFrame(all_metrics).T
df = df.round(4)
print('\n' + '='*70)
print('  SOLUTION COMPARISON')
print('='*70)
print(df.to_string())
```

```python
# Visualization
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Recall metrics
recall_cols = [c for c in df.columns if c.startswith('Recall')]
if recall_cols:
    df[recall_cols].plot(kind='bar', ax=axes[0])
    axes[0].set_title('Recall@K Comparison')
    axes[0].set_ylabel('Score')
    axes[0].tick_params(axis='x', rotation=30)
    axes[0].legend(fontsize=8)

# Ranking metrics
rank_cols = [c for c in df.columns if c in ('AUC', 'MRR@5') or c.startswith('NDCG')]
if rank_cols:
    df[rank_cols].plot(kind='bar', ax=axes[1])
    axes[1].set_title('Ranking Metrics Comparison')
    axes[1].set_ylabel('Score')
    axes[1].tick_params(axis='x', rotation=30)
    axes[1].legend(fontsize=8)

plt.tight_layout()
plt.savefig('../outputs/results/comparison.png', dpi=150, bbox_inches='tight')
plt.show()
```
